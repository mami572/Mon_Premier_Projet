```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Audio ,display
#--------Parameteres_-------------#
fs =8000 #frs
duration= 2.0 #duree
t= np.linspace(0,duration,int(fs*duration),endpoint =False)

#Generere
note1 =0.5 *np.cos(2 * np.pi * 100 * t)
note2 =0.5 *np.cos(2 * np.pi * 300 * t)
mixte =note1 + note2

#tous les frequences des notes

plt.figure(figsize=(15,10))


plt.subplot(4,1,1)
plt.plot(t[:400],note1[:400],color ='blue')
plt.title("Note 1 seule (100 HZ)")


plt.subplot(4,1,2)
plt.plot(t[:400],note2[:400],color ='red')
plt.title("Note 2 seule (300 HZ)")

S_fff =np.fft.fft(mixte)
ferquence = np.fft.fftfreq(len(mixte),1/fs)

#filter pass bas
S_fff_filtree =S_fff.copy()
S_fff_filtree[np.abs(ferquence)>200]=0
signal_filter = np.fft.ifft(S_fff_filtree).real


#3 traces
# Removed: plt,figsize=(15,10) which caused plt to be reassigned to an int
plt.subplot(4,1,1)
plt.plot(t[:400],note1[:400],color ='blue')
plt.title("Note 1 seule (100 HZ)")

plt.subplot(4,1,2)
plt.plot(t[:400],note2[:400],color ='green')
plt.title("Note 2 seule (300 HZ)")




display(Audio(mixte,rate=fs))
display(Audio(signal_filter,rate=fs))
display(Audio(note1,rate=fs))
display(Audio(note2,rate=fs))

#note DO

#freq 261.6   293.6  329.7   349.6  349.2   392.
```
